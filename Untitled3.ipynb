```python
import numpy as np
import os
import pandas as pd
from astropy.io import fits
from astropy.modeling.functional_models import Gaussian2D, Sersic2D

# Buat folder khusus dataset klasifikasi
os.makedirs('classification_dataset', exist_ok=True)

jumlah_gambar = 1000  # Total gambar yang ingin dibuat (misal 500 bintang, 500 galaksi)
size = 300
x, y = np.meshgrid(np.arange(size), np.arange(size))

labels_data = []

print("Mulai membuat dataset klasifikasi...")

for i in range(jumlah_gambar):
    # Acak label: 0 untuk STAR, 1 untuk GALAXY (peluang 50:50)
    label_class = np.random.choice([0, 1])

    if label_class == 0:
        # --- GENERATE BINTANG SAJA ---
        star_amp = np.random.uniform(3, 10)
        star_x = np.random.uniform(50, 250)
        star_y = np.random.uniform(50, 250)
        # Ukuran bintang (FWHM/stddev) dibuat bervariasi tipis
        stddev = np.random.uniform(2, 5)

        star = Gaussian2D(amplitude=star_amp, x_mean=star_x, y_mean=star_y, x_stddev=stddev, y_stddev=stddev)
        image_data = star(x, y)
        class_name = 'star'

    else:
        # --- GENERATE GALAKSI SAJA ---
        gal_amp = np.random.uniform(1, 5)
        gal_x = np.random.uniform(50, 250)
        gal_y = np.random.uniform(50, 250)
        gal_reff = np.random.uniform(15, 50)  # Galaksi cenderung lebih diffuse/lebar dari bintang
        gal_n = np.random.uniform(1, 5)     # Indeks Sersic
        gal_ellip = np.random.uniform(0.1, 0.7)
        gal_theta = np.random.uniform(0, 2 * np.pi)

        galaxy = Sersic2D(amplitude=gal_amp, r_eff=gal_reff, n=gal_n, x_0=gal_x, y_0=gal_y, theta=gal_theta, ellip=gal_ellip)
        image_data = galaxy(x, y)
        class_name = 'galaxy'

    # --- TAMBAHKAN NOISE (Latar Belakang Teleskop Realistis) ---
    noise = np.random.normal(loc=0.0, scale=0.1, size=(size, size))
    final_image = image_data + noise

    # Simpan Gambar FITS
    filename = f'img_{i:04d}.fits'
    filepath = os.path.join('classification_dataset', filename)
    fits.writeto(filepath, final_image, overwrite=True)

    # Catat Label ke Dataframe
    labels_data.append({
        'filename': filename,
        'label': label_class,       # 0 = Star, 1 = Galaxy
        'class_name': class_name
    })

# Simpan CSV label
df_labels = pd.DataFrame(labels_data)
df_labels.to_csv('classification_labels.csv', index=False)

print("Selesai! Dataset siap digunakan untuk training CNN.")
```
